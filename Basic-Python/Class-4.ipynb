```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
# using employees dataset
data=pd.read_csv("employee.csv",nrows=200)

# Lets take only 200 rows as we are only going to visual it.
print(data.shape)
```

```python
# line chart
plt.plot(data['Department'],data['Age'])
plt.show()
```

```python
# Lets try to aggregate the age using mean in each dept.
data.pivot_table(index="Department",values="Age",aggfunc="mean").plot()
```

```python
# Same Using Groupby.
data.groupby("Department")["Age"].mean().plot()
```

```python
data.groupby("Department")[["TotalWorkingYears","Age"]].mean().plot()
```

```python
# Lets try to aggregate the using mean in each dept
data.pivot_table(index="Department",
                  values=["Age","TotalWorkingYears","HourlyRate"],
                  aggfunc="mean").plot()
```

```python
# using the groupby
data.groupby("Department")[["Age","TotalWorkingYears","HourlyRate"]].mean().plot()
```
