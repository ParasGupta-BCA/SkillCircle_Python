# Missing Value Imputation

```python
import pandas as pd
```

```python
df=pd.read_csv("train.csv")
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
# What can we do if we have null values 
# 1. Drop the rows or columns having null values
# 2. Fill The empty cells with some value
```

```python
# Drop the rows

df.dropna() # Complete data loss
```

```python
df.dropna(axis=1) # Column loss
```

```python
# Fill null values 
# First you have to check the data type of the column you want to fill
# if data type is numerical that you can fill that with mean or median 
# if data type is categorical then you can fill with mode or custom value
```

```python
df["LotFrontage"].isnull().sum()
```

```python
df["LotFrontage"].dtype
```

```python
df["LotFrontage"].agg(["mean","median"])
```

```python
df["LotFrontage"].fillna(df["LotFrontage"].median(),inplace=True)
```

```python
df[df["LotFrontage"]==69]
```

```python
df["GarageType"].isnull().sum()
```

```python
df["GarageType"].dtype
```

```python
df["GarageType"].mode()[0]
```

```python
df["GarageType"].fillna(df["GarageType"].mode()[0],inplace=True)
```

```python
df["GarageQual"].isnull().sum()
```

```python
df["GarageQual"].dtype
```

```python
df["GarageQual"].fillna("NA",inplace=True)
```
