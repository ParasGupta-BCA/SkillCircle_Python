```python
import pandas as pd
```

```python
df=pd.read_csv("employee.csv")
```

```python
# Univariate Analysis
# Bivariate Analysis 
# Multivariate Analysis
```

```python
# Univariate Analysis 
# You Can do this with text columns as value_counts()
# You can do this with numerical columnn as mean,median,mode
```

```python
df["Department"].value_counts()
```

```python
df["BusinessTravel"].value_counts()
```

```python
df["Gender"].value_counts()
```

```python
df["EducationField"].value_counts()
```

```python
# Bivariate Analysis
# df.groupby(categorical)[numerical].mean(),median,max,min
```

```python
# What is the average age of employees in each department

df.groupby("Gender")["Age"].mean()
```

```python
# What is the average MonthlyIncome of employees in each department

df.groupby("Department")["MonthlyIncome"].mean()
```

```python
# What is the average MontlyIncome of employees from each EductationField

df.groupby("EducationField")["MonthlyIncome"].mean()
```

```python
# Count how many male and female are in each department 

df.groupby("Department")["Gender"].value_counts()
```

```python
# Multivariate Analysis 
# What is the average Age of employees in each department and educationfield

df.groupby(["Department","EducationField"])["Age"].mean()
```

```python
df.groupby(["Department","EducationField"])["MonthlyIncome"].mean()
```

```python
df.groupby("Department")[["Age","MonthlyIncome"]].mean()
```

```python
df.groupby(["Department","EducationField"])[["Age","MonthlyIncome"]].mean()
```

```python
df.groupby("Department")["Age"].agg(["mean","max","min","median"])
```

```python
df.groupby(["Department","EducationField"])[["Age","MonthlyIncome"]].agg(["mean","max","min","median"])
```

```python
pd.pivot_table(index=["Department","EducationField"],values="MonthlyIncome",data=df,aggfunc="mean")
```
