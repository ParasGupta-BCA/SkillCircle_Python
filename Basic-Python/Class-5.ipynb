```python
current_palette1 = sns.color_palette()
sns.palplot(current_palette1)
plt.show()
```

### 11. Plotting Univariate Distribution with distplot()

* distplot used for see the normality of the distributed
* condition that less value avaiable higher quantity
* the larger value available in less quantity
* or data is equally distribution 
* if there is asymmetry the skewnees occurs


```python
sns.distplot(tips["total_bill"])
```

```python
sns.distplot(tips["total_bill"],hist=False)
```

### 13. Seaborn - Kernel Density Estimates
Kernel Density Estimation (KDE)

```python
sns.distplot(tips["total_bill"],kde=False)
plt.show()
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips)
plt.show()
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips,kind="hex")
plt.show()
```

```python
sns.jointplot(x="tip",y="total_bill",data=tips,kind="reg")
plt.show()
```
