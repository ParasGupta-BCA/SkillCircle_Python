```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("employee.csv",nrows=200)
```

```python
data["Attrition"].value_counts().plot(kind="bar")
```

```python
data["Gender"].value_counts().plot(kind="bar")
```

```python
plt.figure(figsize=(5,3))
data["Attrition"].value_counts().plot(kind="bar",color="green")

plt.title("Univariate Analysis on Attribition Column Of Employee Dataset")
plt.xlabel("Attrition")
plt.ylabel("Values")
plt.grid()
plt.show()
```

```python
# Lets plot Education vs Montly Rate

plt.figure(figsize=(10,4))
data.pivot_table(
    index="EducationField",
    values=["MonthlyRate"],
    aggfunc="mean"
).plot(kind="bar", color="blue")
plt.title("Average Salary of employees from various Department")
plt.grid()
plt.show()
```

```python
data.groupby("EducationField")["MonthlyRate"].mean().plot(kind="bar")
```
