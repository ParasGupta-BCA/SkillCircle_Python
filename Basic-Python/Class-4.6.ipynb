# Pychart

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
# using employees dataset
data=pd.read_csv("employee.csv",nrows=200)

# Lets take only 200 rows as we are only going to visual it.
print(data.shape)
```

```python
plt.pie(data['BusinessTravel'].value_counts(),
        labels = data['BusinessTravel'].value_counts().index,autopct='%1.2f%%',
        explode = [0,0.2,0.3])
plt.show()
```
