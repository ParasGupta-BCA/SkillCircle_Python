```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel('Sample.xls')
```

```python
plt.scatter(x=df["Sales"],y=df["Profit"])
plt.xlable("Sales")
plt.ylable("Profit")
plt.show()
```

```python
sns.relplot(x="Sales",y="Profit",data=df)
```

```python
df["Order Priority"].value_counts()
```

```python
sns.relplot(x="Sales",y="Profit",data=df,hue="Order Priority")
```

```python
sns.relplot(x="Sales",y="Profit",data=df,hue="Order Priority",style="Order Priority")
```

```python
sns.relplot(x="Sales",y="Profit",data=df,hue="Order Priority",style="Ship Mode")
```

```python
df["Ship Mode"].value_counts()
```

```python
sns.relplot(x="Sales",y="Profit",data=df,size="Discount")
```

```python
sns.relplot(x="Sales",y="Profit",data=df,size="Discount",sizes=(25,200))
```

```python
sns.catplot(x="Ship Mode",y="Profit",data=df,palette="rainbow")
```

```python
sns.catplot(x="Ship Mode",y="Profit",data=df,jitter=False,palette="rainbow")
```

```python
sns.catplot(x="Ship Mode",y="Profit",data=df.query("Profit>1000"),jitter=False)
```

```python
sns.catplot(x="Ship Mode",y="Profit",data=df.query("Profit>1000 & Profit<5000"),kind="swarm",palette="rainbow")
```

```python
sns.countplot(x="Order Priority",data=df,hue="Ship Mode")
```
