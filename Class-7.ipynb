```python
#using break
i=1
while i<5:
    print(i)
    if i==3:
        break
    i += 1
```

```python
i=0
while i<6:
    i += 1
    if i==3:
        continue
    print(i)
```

```python
# else statement
i=1
while i<6:
    print(i)
    i += 1
else:
    print("I is not logger less than 6")
```

```python
import random

number=random.randint(1,100)
attempts=0

print("---The Great Number Guessing---")
print("I Have Choose a number between 1 and 100. can you guess it?")

while True:
    guess=int(input("Enter the number for guess:"))
    attempts += 2

    if(guess< number):
        print("Too low! Try a higher number.")
    elif(guess>number):
        print("Too Low! Try a lowers number.")
    else:
        print(f"Congratulation! You found it in {attempts} attempts.")
        break
```
