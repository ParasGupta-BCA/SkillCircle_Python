#### Conditional Statements
When We want to take the decision on the basis of condition, then we use the condition statements!

#### 1. if 
used when we want to handle a case of true condition only!

```python
print("Today is the voting day!")
print()

age=24
if age>18:
    print("You Can Vote Today!")
print("Done!")
```

```python
print("Today is the voting day! \n")

# it will take input from user on runtime
age = int(input("Enter Your age:"))

if age >= 18:
    print("You are eligible voter,\n go and cast the vote immediately!")
print("Exit!")
```

```python
Marks_1=int(input("Enter Your English Marks:"))
Marks_2=int(input("Enter Your Hindi Marks:"))
Marks_3=int(input("Enter Your Science Marks:"))
Marks_4=int(input("Enter Your Maths Marks:"))

avg=(Marks_1+Marks_2+Marks_3+Marks_4)/4

if avg<=40 :
    print("\nPassed")
else:
    print("Failed")
```

```python
print("Today is the voting day! \n")

# it will take input from user on runtime
age = int(input("Enter Your age:"))

nationality = input("Enter Nationality:").upper()

if age >= 18 and nationality =="Indian":
    print("You are eligible voter,\n go and cast the vote immediately!")
print("Exit!")
```
