#### Matplotlib And Seaborn

```python
import matplotlib.pyplot as plt
import numpy as np

xpoint=np.array([0,6])
ypoint=np.array([0,250])

plt.plot(xpoint,ypoint)
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[5,6,1,3,2]
num2=[20,60,10,2,4]

plt.plot(num1,num2,marker="o",markersize=10,markerfacecolor='green')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[5,6,1,3,2]
num2=[20,60,10,2,4]

plt.plot(num1,num2,marker="o",markersize=10,markerfacecolor='green',linestyle='--')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[5,6,1,3,2]
num2=[20,60,10,2,4]

plt.plot(num1,num2,marker="X",markersize=10,markerfacecolor='blue',linestyle='--',color='red',linewidth=2)
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[5,6,1,3,2]
num2 = [20, 60, 10, 20, 40]

plt.plot(num1,marker="X",markersize=10,markerfacecolor='blue',linestyle='--',color='red',linewidth=2)
plt.plot(num2,marker="o",markersize=7,markerfacecolor='red',linestyle='--',color='blue',linewidth=2)
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[5,6,1,3,2]
num2 = [20, 60, 10, 20, 40]
plt.plot(num1)
plt.plot(num2)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("Line Chart")
plt.legend(["num1","num2"])
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[5,6,1,3,2]
num2=[20,60,10,2,4]

plt.bar(num1,num2)
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[5,6,1,3,2]
num2=[20,60,10,2,4]

plt.barh(num1,num2)
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
num1=[1,2,3,4,5]
num2=[10,20,30,40,50]
plt.pie(num1,labels=["A","B","C","D","E"],colors=["red","green","yellow","blue","purple"])
plt.legend()
plt.show()
```
