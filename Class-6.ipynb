### Join Union In Python

```python
set1={"a","b","c"}
set2={1,2,3}

set3=set1.union(set2)
print(set3)
```

```python
set1={"a","b","c"}
set2={1,2,3}

set3=set1 | set2
print(set3)
```

Multiple Sets In Python

```python
set1={"a","b","c"}
set2={1,2,3}
set3={"John", "Elena"}
set4={"apple","banana","cherry"}

myset=set1.union(set2,set3,set4)
print(myset)
```

```python
set1={"a","b","c"}
set2={1,2,3}
set3={"John", "Elena"}
set4={"apple","banana","cherry"}

myset=set1 | set2 | set3 |set4
print(myset)
```
