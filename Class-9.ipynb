### Lambda In Python

```python
x=lambda a : a+10
print(x(5))
```

```python
x=lambda a,b : a*b
print(x(5,6))
```

```python
x=lambda a,b,c : a+b+c
print(x(10,20,30))
```

```python
def myfunc(n):
    return lambda a : a*n
mydoubler=myfunc(5)
mytripler=myfunc(10)

mydoubler = myfunc(5)
mytripler = myfunc(10)

print(mydoubler(4))
print(mytripler(40))
```

##### The map() function executes a specified function for each item in an iterable. The item is sent to the function as a parameter.

```python
number =[1,2,3,4,5]
doubled =list(map(lambda x:x*2,number))

print(doubled)
```

##### The filter() function is a built-in Python tool used to extract elements from an iterable (like a list, tuple, or set) that satisfy a specific condition.

```python
number =[1,2,3,4,5]
odd_number =list(filter(lambda x : x%2 !=0,number))

print(odd_number)
```

```python
# Countdown is the function not the build-in function in python

def countdown(n):
    if n<=0:
        print("Done!")
    else:
        print(n)
        countdown(n-1)
countdown(5)
```

#### IMP FOR INTERVIEW - Factorial In Python

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(5))
```
