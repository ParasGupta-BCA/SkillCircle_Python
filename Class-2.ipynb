## Operator
Operators are special symbols that perform actions on values or variables (called operands).

## Arithmetic Operators:

##### 1. (+) → adds numbers or joins strings
##### 2. (-) → subtracts
##### 3. (*) → multiplies
##### 4. (/) → divides
##### 5. (==) → checks if two things are equal

```python
a=45
b=30

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**b)
print(a//b)
print(a%b)
```

### 2. Realtion Operatiors 

##### it returns the true or false, used to compare two value using a sign of inequality or equalities!

```python
a=45
b=20

print(a>b)
print(a<b)
print(b>a)
print(a>=45)
print(b<=20)
print(b<=4)
print(a==b)
print(a==45)
print(a!=b)
```

### 2. Logical Operatiors 

##### when we want to make a combined decision from multiple conditions!

##### and --> return False, Only when atlest 1 is False
##### 1. T or T -> True
##### 2. T or F -> True
##### 3. F or T -> True
##### 1. F or F -> False

```python
(45>30) and (30>20)
```

```python
(45>30) and (30<20)
```

```python
(45<30) and (30>20) and (10>5)
```

##### OR ->
