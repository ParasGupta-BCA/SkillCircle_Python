### Class-5: List In Python

```python
fruits=["apple","banana","cherry","kiwi","mango"]
newlist=[]

for x in fruits:
    if "a" in x:
        newlist.append(x)

print(newlist)
```

```python
thislist=["apple","banana","cherry","kiwi","mango"]
thislist.sort()
print(thislist)
```

```python
thislist=["orange","mango","kiwi","pineapple","banana"]
thislist.sort(reverse=True)
print(thislist)
```

#### Sort the list descending in list.

```python
thislist=[100,50,65,82,23]
thislist.sort(reverse=True)
print(thislist)
```

#### Sort the list Ascending in list.

```python
thislist=[100,50,65,82,23]
thislist.sort(reverse=False)
print(thislist)
```

#### Use the copy() menthod

```python
thislist=["apple","banana","cherry","kiwi","mango"]
mylist=thislist.copy()
print(mylist)
```

#### Use the list() method

```python
thislist=["apple","banana","cherry","kiwi","mango"]
mylist=list(thislist)
print(mylist)
```

```python
list1=["a","b","c"]
list2=[1,2,3]

list=list1 + list2
print(list)
```

### Tuple In Python
It Is Unchangable or Imutable.

```python
thistuple=("apple","banana","cherry")
print(type(thistuple))
print(thistuple)
```

```python
thistuple=("apple","banana","cherry")
print(thistuple[1])
```

```python
thistuple=("apple","banana","cherry")
print(thistuple[-1])
```

```python
thislist=["apple","banana","cherry","orange","kiwi","melon","mango"]
print(thislist[2:5])
```

```python
thislist=["apple","banana","cherry","orange","kiwi","melon","mango"]
print(thislist[2::5])
```

```python
thislist=["apple","banana","cherry","orange","kiwi","melon","mango"]
print(thislist[:4])
```

```python
thislist=["apple","banana","cherry","orange","kiwi","melon","mango"]
print(thislist[2:4])
print(thislist[2::4])
```
